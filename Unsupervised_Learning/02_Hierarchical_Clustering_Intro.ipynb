# Hierarchical Clustering

Hierarchical clustering involves creating clusters that have a predetermined ordering from top to bottom. There are two types: agglomerative (bottom-up) and divisive (top-down).

Here we will use **Agglomerative Clustering**, which is more common. We will also use **Dendrograms** to visualize the hierarchy.


```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from scipy.cluster.hierarchy import dendrogram, linkage
from sklearn.cluster import AgglomerativeClustering

%matplotlib inline
```

## 1. Generating Data


```python
X, y = make_blobs(n_samples=200, centers=3, cluster_std=1.0, random_state=42)
plt.scatter(X[:,0], X[:,1], s=50)
plt.title('Raw Data')
plt.show()
```

## 2. Dendrogram
A dendrogram helps us understand the relationships between the data points and figure out the optimal number of clusters.


```python
linked = linkage(X, 'ward')
plt.figure(figsize=(10, 7))
dendrogram(linked, orientation='top', distance_sort='descending', show_leaf_counts=True)
plt.title('Hierarchical Clustering Dendrogram')
plt.show()
```

## 3. Agglomerative Clustering
Based on the dendrogram, we can choose to cut the tree at a certain height to get clusters. Let's use 3 clusters.


```python
cluster = AgglomerativeClustering(n_clusters=3, metric='euclidean', linkage='ward')
y_pred = cluster.fit_predict(X)

plt.scatter(X[:,0], X[:,1], c=y_pred, cmap='viridis', s=50)
plt.title('Agglomerative Clustering Results')
plt.show()
```
