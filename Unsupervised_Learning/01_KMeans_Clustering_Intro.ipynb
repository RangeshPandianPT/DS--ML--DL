# Introduction to Unsupervised Learning: K-Means Clustering

In supervised learning, we have labels (y) for our data (X). In **Unsupervised Learning**, we only have the data (X) and no labels. The goal is to find hidden structures or patterns in the data.

One of the most common unsupervised learning tasks is **Clustering**. The algorithm groups similar data points together. 

**K-Means** is a popular clustering algorithm. It works by:
1. Initializing K random centroids (cluster centers).
2. Assigning each data point to the nearest centroid.
3. Updating the centroids by calculating the mean of all points assigned to them.
4. Repeating steps 2 and 3 until the centroids no longer move.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans

%matplotlib inline
```

## 1. Generating Synthetic Data
Let's create a simple 2D dataset with 4 distinct clusters.

```python
# Generate 300 samples with 4 centers (clusters)
X, _ = make_blobs(n_samples=300, centers=4, cluster_std=0.60, random_state=0)

# Visualize the data
plt.scatter(X[:, 0], X[:, 1], s=50)
plt.title('Unlabeled Data')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.show()
```

## 2. Applying K-Means Clustering
We will tell K-Means to look for 4 clusters.

```python
# Initialize KMeans with 4 clusters
kmeans = KMeans(n_clusters=4, random_state=0, n_init='auto')

# Fit the model and predict the cluster labels for each data point
y_kmeans = kmeans.fit_predict(X)

print("Cluster assignments for the first 10 points:")
print(y_kmeans[:10])
```

## 3. Visualizing the Clusters
Let's plot the data points colored by their assigned cluster, and also plot the cluster centers.

```python
plt.scatter(X[:, 0], X[:, 1], c=y_kmeans, s=50, cmap='viridis')

centers = kmeans.cluster_centers_
plt.scatter(centers[:, 0], centers[:, 1], c='red', s=200, alpha=0.7, marker='X', label='Centroids')

plt.title('K-Means Clustering Results')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.legend()
plt.show()
```
